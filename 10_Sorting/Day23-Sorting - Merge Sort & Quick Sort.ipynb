```python

def is_sorted(arr):
    sorted = True

    for i in range(len(arr)-1):
        if arr[i]> arr[i+1]:
            sorted = False
    return sorted
```

```python
arr = [6, 2, 5, 3]

def bubble_sort(arr):
    for i in range(len(arr) - 1):
        for j in range(len(arr) - 1 - i):
            print(i, j, arr)
            if arr[j] > arr[j+1]:
                arr[j] , arr[j+1] = arr[j+1], arr[j]
    return arr

bubble_sort(arr)
```

```python
def selection_sort(arr):
    for i in range(len(arr)-1):
        min = i
        for j in range(i+1, len(arr)):
            #print(i,j)
            if arr[j]<arr[min]:
                min = j

        print(f'min index: {min}, arr[min]: {arr[min]}, {arr}') 
        arr[i], arr[min] = arr[min], arr[i]
        print(f'new-array:{arr}')
    return arr

arr = [8, 6, 3, 2]
#arr = [5,1,1,2,0,0]
selection_sort(arr)
```

# Insertion Sort

```python
arr = [8, 6, 3, 2, 7]

# Insertion sort is a sorting algorithm that works by inserting elements into their correct position in a sorted array.

def insertion_sort( arr):
    for i in range(1, len(arr)):
        key = arr[i]
        j = i - 1

        print(arr, key, j)

        # Shift larger elements to the right
        while j >=0 and arr[j] > key:
            arr[j+1] = arr[j]
            j = j - 1

        arr[j+1] = key

    return arr
    
insertion_sort(arr)
```

# Merge Sort

```python
def merge_sort(arr):
    print(arr)  # trace the current segment

    if len(arr) <= 1:
        return arr

    mid = len(arr) // 2
    left = merge_sort(arr[:mid])
    right = merge_sort(arr[mid:])

    return merge(left, right)


def merge(left, right):
    result = []
    i = j = 0

    # merge step
    while i < len(left) and j < len(right):
        if left[i] <= right[j]:
            result.append(left[i])
            i += 1
        else:
            result.append(right[j])
            j += 1

    # append leftovers
    result.extend(left[i:])
    result.extend(right[j:])

    return result

print(merge_sort([5,2,3,1]))
```

```python
class Solution:
    def sortArrayByParity(self, nums: List[int]) -> List[int]:

        left = 0
        right = len(nums) - 1

        while left < right:
            print(nums, left, right)

            if nums[left] %2 ==0:
                print('left num is even')
                left = left + 1

            elif nums[right] %2 ==1:
                print('right is odd num')
                right = right -1

            else:
                nums[right], nums[left] = nums[left], nums[right]
                right -=1
                left +=1
                
        return nums

nums = [3,4,1,5]
#Output: [2,4,3,1]
Solution().sortArrayByParity(nums)
```
