A Binary Search Tree (BST) is a node-based binary tree data structure that maintains its elements in a strictly sorted order to allow fast lookups, insertions, and deletions.

The BST Core RuleFor every single node in the tree:
1. All values in its left subtree must be less than the node's value.
2. All values in its right subtree must be greater than the node's value.
3. All values in the Binary search tree should be unique.

```python
class Solution:
    def insertIntoBST(self, root: Optional[TreeNode], val: int) -> Optional[TreeNode]:
        if root is None:
            return TreeNode(val)
        
        temp = root
        while True:
            if val < temp.val:
                if temp.left is None:
                    temp.left = TreeNode(val)
                    break
                temp = temp.left
            else:
                if temp.right is None:
                    temp.right = TreeNode(val)
                    break
                temp = temp.right

        return root
```

```python
class Solution: #Time LogN (base 2) -> space O(1)
    def searchBST(self, root: Optional[TreeNode], val: int) -> Optional[TreeNode]:
        if root is None:
            return None
        temp = root
        
        while temp is not None:
            if temp.val == val:
                return temp
            elif val < temp.val :
                temp = temp.left
            else:
                temp = temp.right
        return None
```

```python
#https://www.geeksforgeeks.org/problems/implementing-ceil-in-bst/1
class Solution:
    def findCeil(self, root, x):
        ceil = -1
        while root:
            if root.data == x:
                return x
            elif x < root.data:
                ceil = root.data
                root = root.left
            else:
                root = root.right
    
        return ceil
```

```python
#https://www.geeksforgeeks.org/problems/minimum-element-in-bst/1
class Solution:
    def minValue(self, root):
        if root is None:
            return -1
        temp = root
        min_val = root.data

        while temp:
            min_val = temp.data
            temp = temp.left 
        return min_val

# time -> O(LogN)
# Space -> Space -> O(1) (So the extra memory remains constant: 2 variables → constant space → O(1))
```

```python
def minValue(root):
    if root.left is None:
        return root.data
    return minValue(root.left)

#Each recursive call is stored on the call stack. For a tree of height h: Space = O(h)

# Iterative traversal with a few variables → O(1) space.
# Recursive traversal → O(h) space due to the call stack.
```

```python
#https://leetcode.com/problems/range-sum-of-bst/
class Solution:
    def rangeSumBST(self, root: Optional[TreeNode], low: int, high: int) -> int:

        def preorder(root, currsum):
            if root is None:
                return 0

            if low <= root.val <= high:
                currsum = root.val
            else:
                currsum = 0

            return currsum + preorder(root.left, currsum) + preorder(root.right, currsum)

        return preorder(root, currsum = 0)
```
