```python
from collections import Counter

class Solution:
    def findTheDifference(self, s: str, t: str) -> str:
        count = Counter(s)
        
        print(count)

        for char in t:

            if count[char] == 0:
                    return char
            
            count[char] -= 1

        return -1
        


s = "aabb"
t = "ababb"
Solution().findTheDifference(s,t)
```

```python
pattern = "abe"

s = "cat cat dog"

class Solution:
    def wordPattern(self, pattern: str, s: str) -> bool:
        words = s.split()
        if len(pattern) != len(s.split()):
            return False
        char_to_word = {}
        word_to_char = {}

        for ch, word in zip(pattern , words):
            print(ch,word)

            print(char_to_word)

            if ch in char_to_word:
                if char_to_word[ch] != word:
                    return False
            else:
                char_to_word[ch] = word

            print(word_to_char)
            
            if word in word_to_char:
                if word_to_char[word] != ch:
                    return False
            else:
                word_to_char[word] = ch


        return True
Solution().wordPattern(pattern, s)
```

I can be placed before V (5) and X (10) to make 4 and 9. 

X can be placed before L (50) and C (100) to make 40 and 90. 

C can be placed before D (500) and M (1000) to make 400 and 900.


```python
class Solution:
    def romanToInt(self, s: str) -> int:
        roman = {
        "I" :1,
        "V" : 5,
        "X" : 10,
        "L": 50,
        "C" : 100, 
        "D":500, 
        "M":1000 }

        res = 0

        for i, j in zip(s,s[1:]):
            print(roman[i], roman[j])
            if roman[i]<roman[j]:
                res = res - roman[i]
            else:
                res = res + roman[i]
            # print(f'res:{res}')

        return res + roman[s[-1]]

s = "MCDXIVII"
Solution().romanToInt(s)
```

```python
class Solution:
    def isIsomorphic(self, s: str, t: str) -> bool:
        if len(s)!=len(t):
            return False

        map = {}
        map2 = {}

        for char1, char2 in zip(s,t):
            if char1 not in map:
                map[char1] = char2
            else:
                if map[char1] != char2:
                    return False
            print(map)

        for char1, char2 in zip(t,s):
            if char1 not in map2:
                map2[char1] = char2
            else:
                if map2[char1] != char2:
                    return False
            print(map2)
        return True
s = "egg1"
t = "addf"
Solution().isIsomorphic(s,t)
```
