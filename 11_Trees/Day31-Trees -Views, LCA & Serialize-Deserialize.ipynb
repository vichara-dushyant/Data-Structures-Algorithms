```python
from collections import deque
from typing import Optional, List

class TreeNode:
    def __init__(self,data):
        self.data = data
        self.left = None
        self.right = None
```

```python
class Solution:

    def __init__(self):
        self.diameter = 0

    def diameterOfBinaryTree(self, root):

        def height(node):

            if node is None:
                print("Reached None -> return 0")
                return 0
            print(f"\nENTER Node {node.data}")
            left_height = height(node.left)
            print(f"Node {node.data}: left child returned height = {left_height}")
            right_height = height(node.right)
            print(f"Node {node.data}: right child returned height = {right_height}")
            current_diameter = left_height + right_height
            print(f"Node {node.data}: diameter through node = " f"{left_height} + {right_height} = {current_diameter}")
            old_diameter = self.diameter
            self.diameter = max(self.diameter, current_diameter)
            print( f"Node {node.data}: global diameter "f"{old_diameter} -> {self.diameter}")
            current_height = 1 + max(left_height, right_height)
            print(f"EXIT Node {node.data}: "f"return height = 1 + max({left_height}, {right_height}) " f"= {current_height}" )
            return current_height
        height(root)
        print(f"\nFINAL DIAMETER = {self.diameter}")
        return self.diameter
```

```python
tree = Solution()
tree.root = TreeNode('A')
tree.root.left = TreeNode('B')
tree.root.right = TreeNode('C')
tree.root.left.left = TreeNode('D')
# tree.root.left.right = TreeNode('E')
tree.root.right.right = TreeNode('F')
# tree.root.left.left.left = TreeNode('G')
# Tree Structure
#
#         A
#       /   \
#      B     C
#     /       \
#    D         F
#   
#  
#
print(tree.diameterOfBinaryTree(tree.root))
```
