```python
#Pre Order Traversal

class Node:
    def __init__(self,data):
        self.data = data
        self.left = None
        self.right = None

class BinaryTreeTraversal: #Traversal when only root address is known
    
    def __init__(self):
        self.root = None

    def preOrder(self, root): #DLR
        if (root is None):
            return
        print(root.data, end = " ")
        self.preOrder(root.left)
        self.preOrder(root.right)

    def inOrder(self, root): #LDR (Print left data right)
        if root is None:
            return
        self.inOrder(root.left)
        print(root.data, end = " ")
        self.inOrder(root.right)

    def postOrder(self, root): #LRD
        if root is None:
            return

        self.postOrder(root.left)
        self.postOrder(root.right)
        print(root.data, end = " ")
```

```python
if __name__ == "__main__":
    tree = BinaryTreeTraversal()

    tree.root = Node(1)
    tree.root.left = Node(2)
    tree.root.right = Node(3)

    tree.root.left.left = Node(4)
    tree.root.left.right = Node(5)

    tree.root.right.left = Node(6)
    tree.root.right.right = Node(7)

    print('\n Pre Order Traversal (DLR):' ,end = " " )
    tree.preOrder(tree.root)

    print(' \n In Order Traversal (LDR) :' ,end = " " )
    tree.inOrder(tree.root)

    print(" \n Post Order Traversal (LRD) :", end = " ")
    tree.postOrder(tree.root)

#   1
# 2  3
#45  67
```

```python
#Definition for a binary tree node.
from typing import Optional, List
class TreeNode:
    def __init__(self, val=0, left=None, right=None):
        self.val = val
        self.left = left
        self.right = right

class Solution:
    def preorderTraversal(self, root: Optional[TreeNode]) -> List[int]: #DLR
        res = []
        self.preorder(root, res)
        return res

    def preorder(self,node, res): #DLR
        if node is None:
            return

        res.append(node.val)
        self.preorder(node.left, res)
        self.preorder(node.right, res)
```

```python
# Number of Nodes in a Binary Tree (Time Complexity is O(n) ->Every node is visited exactly once.
# Balanced tree: height h=logn Space = O(log n)
# Skewed tree (like a linked list): Space = O(n))

class Solution: #Traversal when only root address is known
    def numofNodes(self, root): #DLR
        if root is None:
            return 0
        left = self.numofNodes(root.left)
        right = self.numofNodes(root.right)
        return 1 + left + right

    def numofLeafNodes(self, root): #given address of root
        if root is None:
            return 0
        if root.left is None and root.right is None:
            return 1
        return self.numofLeafNodes(root.left) + self.numofLeafNodes(root.right)

    def numofNonLeafNodes(self, root):
        if root is None:
            return 0
        if root.left is None and root.right is None:
            return 0
        return 1 + self.numofNonLeafNodes(root.left) + self.numofNonLeafNodes(root.right)

print(Solution().numofNodes(tree.root))
print(Solution().numofLeafNodes(tree.root))
print(Solution().numofNonLeafNodes(tree.root))
```
