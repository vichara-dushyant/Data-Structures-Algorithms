```python
class TreeNode:
    def __init__(self, data):
        self.data = data 
        self.children = []
```

BFS is commonly used on
Binary Trees, 
N-ary Trees, 
Graphs, 
Social networks, 
Road maps, 
Shortest path problems.

```python
from collections import deque

class BFSNaryTree: #Generic NaryTree (core idea is -> Use a queue and visit nodes level by level)
    
    def __init__(self):
        self.root = None

    def traverse(self, root):
        bfs = []

        if root is None:
            return bfs

        queue = deque([])
        queue.append(root)

        while len(queue) != 0:
            level_size = len(queue)
            current_level = []

            for _ in range(len(queue)):
                current = queue.popleft()
                current_level.append(current.data)

                for child in current.children:
                    queue.append(child)

            bfs.append(current_level)

        return bfs
```

```python
tree = BFSNaryTree()
tree.root = TreeNode('A')

b = TreeNode('B')
c = TreeNode('C')
d = TreeNode('D')

tree.root.children = [b,c,d]

e = TreeNode('E')
f = TreeNode('F')
g = TreeNode('G')

b.children = [e,f]
d.children = [g]

print(tree.traverse(tree.root))
```

```python
from typing import Optional, List
```

```python
# Definition for a Node.
class Node:
    def __init__(self, val: Optional[int] = None, children: Optional[List['Node']] = None):
        self.val = val
        self.children = children
        
class Solution:
    def preorder(self, root: 'Node') -> List[int]: #DLR -> DFS
        res = []

        self.preodertraversal(root, res)
        return res 
        
    def preodertraversal(self, root, res):

        if root is None:
            return
            
        # Visit Root    
        res.append(root.val)

        if root.children is not None:
            for child in root.children:
                self.preodertraversal(child,res)


# Time Complexity: O(N)
# Space Complexity: O(H) recursive stack (worst case O(N)).
```

```python
tree = Solution()
tree.root = Node('A')

b = Node('B')
c = Node('C')
d = Node('D')

tree.root.children = [b,c,d]

e = Node('E')
f = Node('F')
g = Node('G')

b.children = [e,f]
d.children = [g]

print(tree.preorder(tree.root))
```

```python
class NaryNode:
    def __init__(self, val = None, children= None):
        self.val = val
        self.children = children
```

```python
class Solution:
    def postorder(self, root: 'Node') -> List[int]: #LRD -> DFS
        res = []

        self.postodertraversal(root, res)
        return res 
        
    def postodertraversal(self, root, res):

        if root is None:
            return
            
        if root.children is not None:
            for child in root.children:
                self.postodertraversal(child,res)

        # Visit Root    
        res.append(root.val)
```

```python
tree = Solution()
tree.root = NaryNode('A')

b = NaryNode('B')
c = NaryNode('C')
d = NaryNode('D')

tree.root.children = [b,c,d]

e = NaryNode('E')
f = NaryNode('F')
g = NaryNode('G')

b.children = [e,f]
d.children = [g]

print(tree.postorder(tree.root))
```

```python
# Definition for a binary tree node.
class Node:
    def __init__(self, val=0, left=None, right=None):
        self.val = val
        self.left = left
        self.right = right

from collections import deque

class Solution:
    def levelOrder(self, root: Optional[TreeNode]) -> List[List[int]]:
        bfs = []
        q = deque([])
        q.append(root)

        while len(q) != 0:
            current_level = []
            size = len(q)

            for _ in range(size):
                current = q.popleft()
                current_level.append(current.val)
                if current.left:
                    q.append(current.left)
                if current.right:
                    q.append(current.right)
            bfs.append(current_level)

        return bfs
```

```python
tree = Solution()

tree.root = Node(1)
tree.root.left = Node(2)
tree.root.right = Node(3)

tree.root.left.left = Node(4)
tree.root.left.right = Node(5)

tree.root.right.left = Node(6)
tree.root.right.left.left = Node(7)

tree.levelOrder(tree.root)

#   1
# 2  3
#45  67
```
