# General Tree BFS (N-ary Tree)

```python
class TreeNode:
    def __init__(self, data):
        self.data = data 
        self.children = []
```

```python
#         A
#       / | \
#      B  C  D
#     / \    |
#    E   F   G
```

BFS is commonly used on
Binary Trees, 
N-ary Trees, 
Graphs, 
Social networks, 
Road maps, 
Shortest path problems.

```python
from collections import deque

class BFSNaryTree: #Generic NaryTree (core idea is -> Use a queue and visit nodes level by level)
    
    def __init__(self):
        self.root = None

    def traverse(self, root):
        bfs = []

        if root is None:
            return bfs

        queue = deque([])
        queue.append(root)

        while len(queue) != 0:
            level_size = len(queue)
            current_level = []

            for _ in range(len(queue)):
                current = queue.popleft()
                current_level.append(current.data)

                for child in current.children:
                    queue.append(child)

            bfs.append(current_level)

        return bfs
```

```python
tree = BFSNaryTree()
tree.root = TreeNode('A')

b = TreeNode('B')
c = TreeNode('C')
d = TreeNode('D')

tree.root.children = [b,c,d]

e = TreeNode('E')
f = TreeNode('F')
g = TreeNode('G')

b.children = [e,f]
d.children = [g]

print(tree.traverse(tree.root))
```

```python
from typing import Optional, List
```

```python
# Definition for a Node.
class Node:
    def __init__(self, val: Optional[int] = None, children: Optional[List['Node']] = None):
        self.val = val
        self.children = children
        
class Solution:
    def preorder(self, root: 'Node') -> List[int]: #DLR -> DFS
        res = []

        self.preodertraversal(root, res)
        return res 
        
    def preodertraversal(self, root, res):

        if root is None:
            return
            
        # Visit Root    
        res.append(root.val)

        if root.children is not None:
            for child in root.children:
                self.preodertraversal(child,res)


# Time Complexity: O(N)
# Space Complexity: O(H) recursive stack (worst case O(N)).
```

```python
tree = Solution()
tree.root = Node('A')

b = Node('B')
c = Node('C')
d = Node('D')

tree.root.children = [b,c,d]

e = Node('E')
f = Node('F')
g = Node('G')

b.children = [e,f]
d.children = [g]

print(tree.preorder(tree.root))
```
