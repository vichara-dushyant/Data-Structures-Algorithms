```python
class BinaryTreePathSum:

    def __init__(self):
        self.root = None 
    
    def hasPath(self, root, sum):

        if root is None:
            return False

        if root.data == sum and root.left is None and root.right is None:
            return True 

        return self.hasPath( root.left , sum - root.data) or self.hasPath(root.right, sum - root.data )
        
tree = BinaryTreePathSum()
tree.root = Node(12) 
tree.root.left = Node(7)
tree.root.left.left = Node(9)
tree.root.right = Node(1)
tree.root.right.left = Node(10)
tree.root.right.right = Node(5)

tree.hasPath(tree.root, 18)
```

```python
class Node:
    def __init__(self, data):
        self.data = data
        self.left = self.right = None
```

```python
class BinaryTreePathSum:
    def __init__(self):
        self.root = None 

    def pathsumToleaf(self, root, pathsum):
        if root is None:
            return 0

        pathsum = 10*pathsum + root.data 

        if root.left is None and root.right is None:
            print(f'returing pathsum: {pathsum}')
            return pathsum 

        return self.pathsumToleaf(root.left, pathsum ) + self.pathsumToleaf(root.right, pathsum)

    def findSumPathNumbers(self, root):
        return self.pathsumToleaf(root, 0)
```
