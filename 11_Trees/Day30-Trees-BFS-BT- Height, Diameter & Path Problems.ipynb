```python
# Minimum depth of a Binary Tree
# Consider all root to leaf paths -> then shortest path is minimum

#Traverse using BFS and there will be one Node which is leaf node; that level is the min depth.
```

```python
class Node:
    def __init__(self,data):
        self.data = data
        self.left = None
        self.right = None
```

```python
from collections import deque
```

```python
class BinaryTree:
    def __init__(self):
        self.root = None

    def minimumdepthBFS(self, root):
        bfs = []
        if root is None:
            return 0

        q = deque([])
        q.append(root)

        min_depth = 0
        while len(q) != 0:
            min_depth += 1
            level_size = len(q)
            current_level = []
            
            for _ in range(level_size):
                current = q.popleft()
                
                print(f'Depth: {min_depth} at width/level {level_size} and CurrentData: {current.data}')

                current_level.append(current.data)

                if current.left is not None:
                    q.append(current.left)
                if current.right is not None:
                    q.append(current.right)

                if current.left is None and current.right is None:
                    return min_depth
                    # break
            bfs.append(current_level)
        return bfs
```

```python
    #       6        (min depth = 3) - but len(q) =4
    #     /   \
    #    4     3
    #   / \   / \
    #  2   1 7  5
    #     /             Time: O(n) (worst case visit every node)
    #    8              Space: O(w) where w is the maximum width of the tree
```

```python
tree = BinaryTree()
tree.root = Node(6)
tree.root.left = Node(4)
tree.root.right = Node(3)

tree.root.left.left = Node(2)
tree.root.left.right = Node(1)

tree.root.right.left = Node(7)
tree.root.right.right = Node(5)

tree.root.left.right.left = Node(8)

tree.minimumdepthBFS(tree.root)
```

```python
class TreeNode:
    def __init__(self, data):
        self.data = data 
        self.left = None
        self.right = None
```

```python
from typing import Optional, List

class Solution:
    def maxDepth(self, root: Optional[TreeNode]) -> int:
        if root is None:
            return 0

        q = deque([])
        q.append(root)

        max_depth = 0
        while len(q) != 0:
            max_depth += 1
            level_size = len(q)

            count = 0
            
            for _ in range(level_size):
                current = q.popleft()

                if current.left is not None:
                    q.append(current.left)
                if current.right is not None:
                    q.append(current.right)

                if current.left is None and current.right is None:
                    count +=1

            if count == level_size:
                return max_depth
```

```python
    #       6        (max depth = 4) - but len(q) = 1
    #     /   \
    #    4     3
    #   / \   / \
    #  2   1 7  5
    #     /             Time: O(n) (worst case visit every node)
    #    8              Space: O(w) where w is the maximum width of the tree
```

```python
tree = Solution()
tree.root = TreeNode(6)
tree.root.left = TreeNode(4)
tree.root.right = TreeNode(3)

tree.root.left.left = TreeNode(2)
tree.root.left.right = TreeNode(1)

tree.root.right.left = TreeNode(7)
tree.root.right.right = TreeNode(5)

tree.root.left.right.left = TreeNode(8)

tree.maxDepth(tree.root)
```

```python
class BinaryTree:
    def __init__(self):
        self.root = None

    def rightviewofBT(self, root):
        bfs = []
        if root is None:
            return bfs
        
        q = deque([])
        q.append(root)

        while len(q) != 0:
            level_size = len(q)
            current_level = []

            for _ in range(level_size):
                current = q.popleft()
                
                # if i == level_size - 1: -> Also, we can use this code.
                # res.append(current.data) 

                current_level.append(current.data)

                if current.left is not None:
                    q.append(current.left)
                if current.right is not None:
                    q.append(current.right)

            bfs.append(current_level[-1])
        return bfs
```

```python
tree = BinaryTree()
tree.root = TreeNode(6)

tree.root.left = TreeNode(4)
tree.root.right = TreeNode(3)

tree.root.left.left = TreeNode(2)
tree.root.left.right = TreeNode(1)
tree.root.right.left = TreeNode(7)
tree.root.right.right = TreeNode(5)

tree.root.left.right.left = TreeNode(8)

# Time Complexity: O(n)
# Space Complexity: O(w) where w is the maximum width of the tree (worst case O(n)).

tree.rightviewofBT(tree.root)
```
