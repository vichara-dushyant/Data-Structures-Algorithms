```python
#For Problems like Breadth First Search; we use queue Data Structure
#Start by pushing root node in the queue then find size of queue

# Start with the root node in a queue.
# Process all nodes at the current level.
# Add their children to the queue.
# Repeat until the queue is empty.
# from collections import deque

class Node:
    def __init__(self, data, left = None, right = None):
        self.data = data
        self.left = left
        self.right = right
```

```python
class BreadthFirstSearch: #Level Order Traversal
    def __init__(self):
        self.root = None
        
    def traverse(self, root):
        bfs = []
        if root is None:
            return bfs

        queue = deque([])
        queue.append(root)    
        while len(queue) != 0:
            # print(f'current queue: {[node.data for node in queue]}')
            level_size = len(queue)
            # print(f'size_curr_level: {level_size}')
            current_level = []

            for _ in range(level_size):
                current = queue.popleft()
                current_level.append(current.data)

                if current.left is not None:
                    queue.append(current.left)
                    print(f'Adding left & right of {current.data} in the queue, ->> {current.left.data} , {current.right.data}')

                if current.right is not None:
                    queue.append(current.right)

            bfs.append(current_level)
        return bfs
```

```python
from collections import deque

class BFSNaryTree: #Generic NaryTree (core idea is -> Use a queue and visit nodes level by level)
    
    def __init__(self):
        self.root = None

    def traverse(self, root):
        bfs = []

        if root is None:
            return bfs

        queue = deque([])
        queue.append(root)

        while len(queue) != 0:
            level_size = len(queue)
            current_level = []

            for _ in range(len(queue)):
                current = queue.popleft()
                current_level.append(current.data)

                for child in current.children:
                    queue.append(child)

            bfs.append(current_level)

        return bfs
```

```python
tree = BFSNaryTree()
tree.root = TreeNode('A')

b = TreeNode('B')
c = TreeNode('C')
d = TreeNode('D')

tree.root.children = [b,c,d]

e = TreeNode('E')
# f = TreeNode('F')
g = TreeNode('G')

b.children = [e]
d.children = [g]

print(tree.traverse(tree.root))
```

```python
#ZigZagTraverseInBinaryTree
class ZigZagBFS:
    def __init__(self):
        self.root = None

    def traverse(self, root):
        bfs = []
        if root is None:
            return bfs

        left2right = True
        
        q = deque([])
        q.append(root)
        while len(q) != 0:
            level_size = len(q)
            current_level = deque()

            for _ in range(level_size):
                current = q.popleft()
                if left2right:
                    current_level.append(current.data)
                else:
                    current_level.appendleft(current.data)

                if current.left is not None:
                    q.append(current.left)
                if current.right is not None:
                    q.append(current.right)
                
            bfs.append(list(current_level))
            left2right = not left2right
        return bfs
```

```python
from collections import *

class LevelOrderSuccessor: #Level Order Successor is the node that appears right after the given node in the level order traversal.

    def __init__(self):
        self.root = None

    def key_levelordersuccessor(self, root, key): #For that key which node is level order successor?

        if root is None:
            return None

        q = deque([])
        q.append(root)

        while len(q) != 0:

            current = q.popleft()
            if current.left is not None:
                q.append(current.left)

            if current.right is not None:
                q.append(current.right)

            if current.data == key:
                break

        return q.popleft()
```

```python
tree = LevelOrderSuccessor()
tree.root = Node(2)
tree.root.left = Node(9)
tree.root.right = Node(-7)
tree.root.left.left = Node(31)
tree.root.left.right = Node(27)

# tree.root.right.left = Node(90)
tree.root.right.right = Node(-3)

ans = tree.key_levelordersuccessor(tree.root, 2)
ans.data
```

```python
 from collections import deque

class Solution:
    def leftView(self, root):
        left_view = []

        if root is None:
            return left_view

        q = deque([])
        q.append(root)

        while len(q) != 0:
            level_size = len(q)

            for i in range(level_size):
                current = q.popleft()

                if i == 0:  # First node of the level
                    left_view.append(current.data)

                if current.left is not None:
                    q.append(current.left)

                if current.right is not None:
                    q.append(current.right)

        return left_view
```
