Backtracking is an algorithmic technique that solves problems incrementally by building a candidate solution step-by-step. If at any point it realizes that the candidate cannot lead to a valid solution, it abandons it and reverses ("backtracks") to try the next viable option.

Ex: finding route in a maze

```python
class Solution:
    def subsets(self, nums: List[int]) -> List[List[int]]:
        res = []
        subset = []

        def create_subset(i):
            if i == len(nums):
                res.append(subset[:])
                return
            
            subset.append(nums[i])
            create_subset(i+1)

            subset.pop()
            create_subset(i+1)

        create_subset(0)
        return res
```

```python
class Solution:
    def subsets(self, nums):
        result = []

        def backtrack(index, subset):
            print(f"Running backtrack for index={index}, subset={subset}")

            if index == len(nums):
                print("Adding:", subset)
                result.append(subset[:])
                return

            # Include
            subset.append(nums[index])
            print(f"Include nums[index]: {nums[index]}")
            backtrack(index + 1, subset)

            # Backtrack
            removed = subset.pop()
            print(f"Removed {removed}")

            # Exclude
            #print(f"Exclude nums[index]: {nums[index]}")
            backtrack(index + 1, subset)

        backtrack(0, [])
        return result

Solution().subsets([5,6])

# For every element, you have 2 choices: Include it in the current subset or Exclude it from the current subset.
# This creates a decision tree.
```

```python
class Solution:
    def letterCasePermutation(self, s: str):
        result = []

        def backtrack(index, path):
            if index == len(s):
                result.append("".join(path))
                return

            if s[index].isdigit():
                path.append(s[index])
                backtrack(index + 1, path)
                path.pop()
            else:
                # lowercase
                path.append(s[index].lower())
                backtrack(index + 1, path)
                path.pop()

                # uppercase
                path.append(s[index].upper())
                backtrack(index + 1, path)
                path.pop()

        backtrack(0, [])
        return result

Solution().letterCasePermutation("a1b2")
```

```python
class Solution:
    def subsetSums(self, arr):
        result = []

        def backtrack(index, curr_sum):
            print(f"Running backtrack(index={index}, curr_sum={curr_sum})")

            if index == len(arr):
                print("Adding:", curr_sum)
                result.append(curr_sum)
                return

            # Include current element
            print(f"Include {arr[index]}")
            backtrack(index + 1, curr_sum + arr[index])

            # Exclude current element
            print(f"Exclude {arr[index]}")
            backtrack(index + 1, curr_sum)

        backtrack(0, 0)
        return result


print(Solution().subsetSums([5, 6]))
```
