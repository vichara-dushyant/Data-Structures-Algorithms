```python
# Dynamic Programming -> For all questions
# 1. Recursion(top down) -> TC 2^n ;  SC -> O(N) - stack space 
# 2. Memotization(top down) -> TC = N, SC= N(Stack space) + N (Array) -> To store answers or overlapping subproblems
# 3. Tabulation (Bottom up) -> N, N(reduce space) -> no stack space as no recursion
# 4. Tabulation (Space optimization) -> N, 1
```

```python
#1. Recursion
def fib(num): 
    if num<=1:
        return num
    return fib(num-1) + fib(num-2)

fib(13)
```

```python
#2. Memotization (top down)
num = 5
dp = [-1] * (num + 1)

def fib(num, dp): #Memotization
    if num == 0 or num ==1:
        return num

    if dp[num]!= -1:
        return dp[num]

    dp[num] = fib(num-1, dp) + fib(num-2, dp)
    return dp[num]

fib(num,dp)
```

```python
#3. Tabulation (Bottom up) -> you must know base condition
#[0,1,1,2,3,5,8,13,21] 
#index = 0,1,2,3,4,5,6,7,8

def fib(num):
    dp = [-1]*(num+1)
    dp[0] = 0
    dp[1] = 1

    for i in range(2, num+1):
        dp[i] = dp[i-1] + dp[i-2]
        print(i, dp)

    return dp[num]

fib(5)
```

```python
#4. Tabulation (Space-Optimization) Most Important -> TC -> O(N), SC -> O(1)

def fib(num): #We need only two previous indexes; no need to build a long array.
    prev2 = 0
    prev = 1
    for i in range(2, num+1):
        curr = prev2 + prev

        print(prev2, prev, curr)
        prev2 = prev
        prev = curr
    print(f'curr:{curr}, prev:{prev}')
    return prev #always return prev; curr might go ahead in some cases, here curr is same as prev

fib(9)
```

```python
def climbing_stairs(num): # either 1 or 2 stairs can be climbed at once
    if num<=1:
        return num
    prev2 = 1
    prev = 2

    for i in range(2, num):
        curr = prev + prev2
        prev2 = prev
        prev = curr

    return prev

# 1,2,3,5,8,13,21
climbing_stairs(7)
```

```python
#https://www.geeksforgeeks.org/problems/geek-jump/1

h = [20, 30, 40, 20] #-> func(3) = min( func(2) + abs(h[3]-h[2]) , func(1) + abs(h[3]-h[1])
#0,1,2,3
#Output: 20

def func(index): #Recursion - TC: 2^n, space O(N)
    if index ==0:
        return 0

    jump1 = func(index-1) + abs(h[index-1] - h[index])
    if index >1:
        jump2 = func(index-2) + abs(h[index-2] - h[index-1])
    else:
        jump2 = float('inf')

    return min(jump1, jump2)

func(3)
```

```python
# 2. Memotization (Top Down)
dp = [-1]*len(h)

def frogJump(index, dp):

    if index == 0:
        return 0

    if dp[index] != -1:
        return dp[index]

    jump1 = frogJump(index-1, dp) + abs(h[index-1] - h[index])
    if index >1:
        jump2 = frogJump(index-2, dp) + abs(h[index-2] - h[index])
    else:
        jump2 = float('inf')

    dp[index] = min(jump1,jump2)
    return dp[index]
```

```python
# 3. Tabulation (Bottom Up)
h = [20, 30, 40, 20]
# 0, 1, 2, 3
n = 4
def frogJump(n,h):
    dp = [-1]*len(h)
    dp[0] = 0

    for i in range(1,n):
        jump1 = dp[i-1] + abs(h[i] - h[i-1])

        if i>1:
            jump2 = dp[i-2] + abs(h[i-2] - h[i])
        else:
            jump2 = float('inf')

        dp[i] = min(jump1, jump2)
        print(f'dp:{dp}')
    return dp[n-1]

frogJump(n,h) # TC: N, SC: N
```

```python
nums = [2,7,9,3,1]
# 0,1,2,3,4
#Output: 12
def HouseRobber(index): #1. Recursion  - top down

    if index == 0:
        return nums[index]

    if index <0:
        return 0

    pick = nums[index] + HouseRobber(index-2)
    not_pick = 0 + HouseRobber(index-1)

    return max(pick, not_pick)

# Time Complexity -> Time Limit Exceeded
# O(2^N) (many repeated subproblems)
# Space Complexity
# O(N) (recursion stack space)
```

```python
class Solution:
    def rob(self, nums: List[int]) -> int:
        dp = [-1]*len(nums)
        return self.HouseRobber( len(nums) -1, nums, dp)

    
    def HouseRobber(self, index, nums, dp): #Memotization (top down)
        
        if index == 0:
            return nums[index]
        if index <0:
            return 0

        if dp[index]!=-1:
            return dp[index]

        pick = nums[index] + self.HouseRobber(index-2, nums, dp)
        not_pick = 0 + self.HouseRobber(index-1, nums, dp)

        dp[index] = max(pick, not_pick)
        
        print(f'dp:{dp}, index:{index}')

        return dp[index]

nums = [2,7,9,3,99,100]

Solution().rob(nums)

# TC - O(N)
# SC - O(N) + O(N)
```

```python
nums = [2,7,9,3,99,100]
def func(nums): #Tabulation (Bottom up)
    n = len(nums)
    dp = [-1]*n

    dp[0] = nums[0]
    for index in range(1,n):
        if index >1:
            pick = nums[index] + dp[index-2]
        else:
            pick = nums[index]

        notpick = 0 + dp[index-1]
        dp[index] = max(pick, notpick)
        print(f'dp:{dp}')
    return dp[n-1]

# TC - O(N), Space - O(N)
func(nums)
```
