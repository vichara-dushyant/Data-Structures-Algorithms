```python
# 2. Memoization
class Solution:
    def canPartition(self, nums: List[int]) -> bool:
        total = sum(nums)

        if total % 2 == 1:
            return False

        target = total // 2
        n = len(nums)

        # -1 = not computed
        dp = [[-1 for _ in range(target + 1)] for _ in range(n)]

        return self.func(n - 1, target, nums, dp)

    def func(self, index, target, nums, dp):

        # Target achieved
        if target == 0:
            return True

        # Only one element left
        if index == 0:
            return nums[0] == target

        # Memoization
        if dp[index][target] != -1:
            return dp[index][target]

        # Pick
        if nums[index] <= target:
            pick = self.func(index - 1, target - nums[index], nums, dp)
        else:
            pick = False

        # Not Pick
        not_pick = self.func(index - 1, target, nums, dp)

        dp[index][target] = pick or not_pick
        return dp[index][target]
```

```python
# 4. Tabulation with space optimization
class Solution:
    def canPartition(self, nums: List[int]) -> bool:
        n = len(nums)
        target = sum(nums)

        if target % 2 ==1:
            return False 

        k = target//2 #equal sum for both sub arrays
        prev = [False for _ in range(k+1)]
        prev[0] = True

        if nums[0]<=k:
            prev[nums[0]] = True
        
        for index in range(1,n):
            curr = [False for _ in range(k+1)]

            for target in range(k+1):
                if nums[index]>target:
                    pick = False
                else:
                    pick = prev[target-nums[index]]
                
                not_pick = prev[target]
                curr[target] = pick or not_pick

            prev = curr 
        if prev[k] ==True:
            return True
        else:
            return False
```
