```python
class Solution:
    def longestConsecutive(self, nums: List[int]) -> int:
        if not nums:
            return 0
        temp = sorted(nums)
        print(temp)
        max_len = curr = 1
        for i in range(len(temp)-1):
            if temp[i] == temp[i+1]:
                continue 
            elif temp[i] + 1 == temp[i+1]:
                curr = curr + 1
            else:
                curr = 1
            max_len = max(max_len, curr) 
        return max_len
```

```python
from typing import List

class Solution:
    def letterCombinations(self, digits: str) -> List[str]:

        if not digits:
            return []

        phone = {
            '2': 'abc',
            '3': 'def',
            '4': 'ghi',
            '5': 'jkl',
            '6': 'mno',
            '7': 'pqrs',
            '8': 'tuv',
            '9': 'wxyz'
        }

        result = []

        def backtrack(index, path):
            # Base case: one letter chosen for every digit
            if index == len(digits):
                result.append(path)
                return

            # Get letters corresponding to current digit
            letters = phone[digits[index]]

            for ch in letters:
                backtrack(index + 1, path + ch)

        backtrack(0, "")
        return result

Solution().letterCombinations(digits)
```

```python
def n_fatorial(num):

    if num == 0 or num == 1:
        return 1
        
    return num * n_fatorial(num-1)

n_fatorial(3)
```

```python
class Solution:
    def fib(self, n):

        ans = [0,1] #DP

        for i in range(2, n+1):
            ans.append(ans[i-1] + ans[i-2])
            print(ans)
```
