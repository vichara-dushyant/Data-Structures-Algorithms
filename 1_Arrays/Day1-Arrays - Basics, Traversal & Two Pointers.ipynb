```python
#Array contain duplicates
nums = [1,2,3,1,8,0]

from typing import List

def contain_duplicates(nums: List[int]):
    seen = {}
    for i in range(len(nums)):
        print(seen)
        if nums[i]  in seen:
            return True
        else:
            seen[nums[i]] = 1
    
    return False

print(contain_duplicates(nums))
```

```python
# Two Sum

nums = [2,7,11,15]
target = 18

def twosum(nums : List[int], target):
    seen = {}

    for i in range(len(nums)):
        val = target - nums[i]

        if val in seen:
            return [ seen[val], i]
        else:
            seen[ nums[i]] = i

twosum(nums, target)
```

```python
# Missing Number

class Solution:
    def missingNumber(self, nums: List[int]) -> int:

        total = sum(nums)
        n = len(nums)
        actual = n*(n+1)//2

        return actual - total

Solution().missingNumber([0,1,2,3,4,5,6,9,11,8,12,7])
```

```python
# Find Disapperaed Numbers in an array [1,n]

nums = [4,3,2,7,8,2,3,1]

#nums = [1,1,2]


def findDisappearedNumbers(nums):

    nums_set = set(nums)
    print(nums_set)

    return [ i for i in range(1, len(nums) +1 ) if i not in nums_set ]

findDisappearedNumbers(nums)
```

```python
#Majority Element

nums = [2,2,1,1,1,2,2]

def majorityElement( nums: List[int]) -> int:
        seen = {}
        for i in range(len(nums)):
            if nums[i] not in seen:
                seen[nums[i]] = 1
            else:
                seen[nums[i]] += 1

        for key, value in seen.items():

            if value > len(nums)//2:
                return key
                

majorityElement(nums)
```
