```python
class LinkedList:
    def __init__(self):
        self.head = None
        self.n = 0

    def __len__(self): #DunderMethod
        return self.n

#INSERT
    def insert_head(self,val):
        new_node = Node(val)
        new_node.next = self.head
        self.head = new_node
        self.n = self.n + 1

    def __str__(self): #print
        curr = self.head
        result = ''
        while curr is not None:
            result = result +  str(curr.data) + '->' 
            curr = curr.next
        return result + "None"

    def traverse(self):
        curr = self.head
        while curr:
            print(curr.data, end = '->')
            curr = curr.next
    
    def append(self, val): #insert from tail
        new_node = Node(val)
        curr = self.head
        while curr.next != None:
            curr = curr.next
            
        curr.next = new_node
        new_node.next = None
        self.n = self.n + 1

##CLEAR
    def clear(self):
        self.head = None
        self.n = 0

    def delete_head(self):
        if self.head == None:
            return "Empty LL"
        curr = self.head
        self.head = curr.next
        self.n = self.n - 1

    def pop(self): #delete from tail
        curr = self.head
        while curr.next.next is not None:
            curr = curr.next
        print(curr.data)
        curr.next = None
        self.n = self.n - 1

## SEARCH
    def search(self, item): #return index (search by value)
        pos = 0
        curr = self.head
        while curr:
            if curr.data ==item:
                print(f'Found at pos: {pos}')
                return pos
            pos = pos + 1    
            curr = curr.next
        return 'Not Found'

    def __getitem__(self, index): #return value (search by index)
        pos = 0
        curr = self.head

        while curr is not None:
            if pos ==index:
                return curr.data
            curr = curr.next
            pos = pos+ 1

        return "IndexError" #run using ll[2]

    def reverse(self):
        prev = None
        curr = self.head
        while curr is not None:
            next = curr.next
            curr.next = prev
            prev = curr
            curr = next
        self.head = prev
```

```python
from typing import Optional

class ListNode:
    def __init__(self, val=0, next=None):
        self.val = val
        self.next = next


class Solution:
    def reverseList(self, start: Optional[ListNode]) -> Optional[ListNode]:

        prev = None
        curr = start

        while curr:
            next_node = curr.next
            curr.next = prev

            prev = curr
            curr = next_node

        return prev


def print_list(start):
    curr = start
    while curr:
        print(curr.val, end=" -> ")
        curr = curr.next
    print("None")


# Create linked list
start = ListNode(2)
start.next = ListNode(3)
start.next.next = ListNode(5)
start.next.next.next = ListNode(6)

print("Original List:")
print_list(start)

reversed_start = Solution().reverseList(start)

print("Reversed List:")
print_list(reversed_start)
```

```python
class Solution:
    def replace_max(self, start: Optional[ListNode], new_digit) -> int:

        max_val = start.val
        max_node = start

        curr = start

        while curr is not None:
            if curr.val > max_val:
                max_val = curr.val
                max_node = curr
            curr = curr.next

        max_node.val = new_digit

        return start

max_replaced = Solution().replace_max(start, 64)
print_list(max_replaced)
```

```python
class Solution:
    def remove_duplicates(self, start: Optional[ListNode]) -> int:

        hash_set = set()
        curr = start
        prev = None
        while curr is not None:

            if curr.val in hash_set:
                prev.next = curr.next
                curr = curr.next
            else:

                hash_set.add(curr.val)
                prev = curr
                curr = curr.next

        return start

removed = Solution().remove_duplicates(start)
print_list(removed)
```
