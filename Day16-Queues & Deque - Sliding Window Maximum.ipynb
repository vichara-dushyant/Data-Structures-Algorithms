```python
class ListNode:
    def __init__(self, val, next = None):
        self.val = val
        self.next = next
```

```python
class Queue:

    def __init__(self):
        self.front = None
        self.rear = None

    # Two pointers       Front
    #                    ↓
    #                    1 -> 2 -> 3 -> 4 -> 5
    #                                   ↑
    #                                   Rear
    def enqueue(self, val):
        new_node = ListNode(val)

        if self.rear ==None:
            self.front = self.rear = new_node       
        else:
            self.rear.next = new_node
            self.rear = new_node

    def isempty(self):
        return self.rear == None

    def traverse(self):
        curr = self.front
        print(f"Front: {self.front.val}")
        print(f"Rear: {self.rear.val}")

        while curr:
            print( str(curr.val) , end = "->")
            curr = curr.next

    def dequeue(self):
        if self.isempty():
            return 'empty queue'
        else:
            self.front = self.front.next

    def size(self):
        size = 0
        if self.isempty():
            return size
        else:
            curr = self.front
            while curr:
                curr = curr.next
                size = size + 1
            return size

    def front_item(self):
        if self.front == None:
            return "Empty Queue"
        else:
            return self.front.val

    def rear_item(self):
        if self.rear == None:
            return "Empty Queue"
        else:
            return self.rear.val


q = Queue()
print(q.isempty())
q.enqueue(1)
q.enqueue(2)
q.enqueue(3)
# q.enqueue(4)
#q.dequeue()

q.traverse()
q.size()

# 1.next → 2
# 2.next → 3
# 3.next → 4
```

```python
from collections import deque

class MyStack:

    def __init__(self):
        self.q = deque()

    def push(self, x: int) -> None:
        self.q.append(x)
        print(self.q)
        for _ in range(len(self.q) - 1):
            print(f'bef popleft' , self.q)
            self.q.append(self.q.popleft())
        print(f'after:' , self.q)     

    def pop(self) -> int:
        return self.q.popleft()

    def top(self) -> int:
        return self.q[0]

    def empty(self) -> bool:
        return len(self.q) == 0

    def __str__(self):
        return f"Stack(top->bottom): {list(self.q)}"

obj = MyStack()
```

```python
from typing import List
class Solution:
    def maxSlidingWindow(self, nums: List[int], k: int) -> List[int]:

        result = []

        for i in range(len(nums) - k + 1):

            # print(i)

            num = max(nums[i:i+k])

            result.append(num)

        return result

Solution().maxSlidingWindow(nums,k)
```

```python
from collections import deque

class Solution:
    def maxSlidingWindow(self, nums:List[int], k:int) -> List[int]:
        dq = deque()
        result = []

        for i in range(len(nums)):
            # Remove indices outside current window
            while dq and dq[0] <= i - k:
                dq.popleft()

            #Remove smaller elements
            while dq and nums[dq[-1]] < nums[i]:
                dq.pop()
            dq.append(i)

            # Window formed
            if i>= k-1:
                result.append(nums[dq[0]])
                
        return result

Solution().maxSlidingWindow(nums,k) #Time: O(n) ✅ Space: O(k)
```

```python
from collections import deque

class RecentCounter:

    def __init__(self):
        self.q = deque()

    def ping(self, t: int) -> int:

        self.q.append(t)

        while self.q[0] < t - 3000:
            print(self.q[0], "popleft")
            self.q.popleft()

        return f'length: {len(self.q)}'

    def __str__(self):
        return f"{list(self.q)}"

# Your RecentCounter object will be instantiated and called as such:
# obj = RecentCounter()
# param_1 = obj.ping(t)

recentCounter  = RecentCounter()
recentCounter.ping(1)
recentCounter.ping(10)
```
