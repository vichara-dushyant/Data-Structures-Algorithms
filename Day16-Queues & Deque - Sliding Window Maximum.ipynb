```python
class ListNode:
    def __init__(self, val, next = None):
        self.val = val
        self.next = next
```

```python
class Queue:

    def __init__(self):
        self.front = None
        self.rear = None

    # Two pointers       Front
    #                    ↓
    #                    1 -> 2 -> 3 -> 4 -> 5
    #                                   ↑
    #                                   Rear
    def enqueue(self, val):
        new_node = ListNode(val)

        if self.rear ==None:
            self.front = self.rear = new_node       
        else:
            self.rear.next = new_node
            self.rear = new_node

    def isempty(self):
        return self.rear == None

    def traverse(self):
        curr = self.front
        print(f"Front: {self.front.val}")
        print(f"Rear: {self.rear.val}")

        while curr:
            print( str(curr.val) , end = "->")
            curr = curr.next

    def dequeue(self):
        if self.isempty():
            return 'empty queue'
        else:
            self.front = self.front.next

    def size(self):
        size = 0
        if self.isempty():
            return size
        else:
            curr = self.front
            while curr:
                curr = curr.next
                size = size + 1
            return size

    def front_item(self):
        if self.front == None:
            return "Empty Queue"
        else:
            return self.front.val

    def rear_item(self):
        if self.rear == None:
            return "Empty Queue"
        else:
            return self.rear.val


q = Queue()
print(q.isempty())
q.enqueue(1)
q.enqueue(2)
q.enqueue(3)
# q.enqueue(4)
#q.dequeue()

q.traverse()
q.size()

# 1.next → 2
# 2.next → 3
# 3.next → 4
```
