```python
# Definition for singly-linked list.
# class ListNode:
#     def __init__(self, val=0, next=None):
#         self.val = val
#         self.next = next
class Solution:
    def middleNode(self, head: Optional[ListNode]) -> Optional[ListNode]:
        curr = head
        leng = 0

        while curr is not None:
            leng = leng + 1
            curr = curr.next

        pos = 0
        start = head
        temp = None

        while start is not None:
            if pos == leng//2 :
                temp = start
            start = start.next
            pos = pos + 1
        head = temp
        return head

#head = [1,2,3,4,5,6]
#Output: [4,5,6]
```

```python
class Solution:
    def middleNode(self, head: Optional[ListNode]) -> Optional[ListNode]:
        slow = head
        fast = head

        while fast and fast.next:
            slow = slow.next
            fast = fast.next.next
        return slow
```

```python
class Solution:
    def detectCycle(self, head: Optional[ListNode]) -> Optional[ListNode]:
        node=head
        seen=set()
        while(node and node not in seen):
            seen.add(node)
            node=node.next
        return node
```

```python
class Solution:
    def getIntersectionNode(self, headA: ListNode, headB: ListNode) -> Optional[ListNode]:

        lenA = 0
        tempA = headA
        while tempA:
            lenA = lenA + 1
            tempA = tempA.next

        lenB = 0
        tempB = headB
        while tempB:
            lenB = lenB + 1
            tempB = tempB.next

        diff = abs(lenA - lenB)

        if lenA > lenB:
            while diff > 0:
                headA = headA.next
                diff = diff -1 
        else:
            while diff> 0:
                headB = headB.next
                diff = diff -1 

        while headA != headB:
            headA = headA.next
            headB = headB.next
        return headA
```
