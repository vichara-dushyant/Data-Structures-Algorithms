```python
## Two sum using two pointers

arr = [2,3,4,5,6]
target= 100
#return 4,6

def twosum(arr, target):

    left = 0
    right = len(arr) - 1    
    
    while left < right:
        print(left, right)
        
        if arr[left] + arr[right] == target:
            return arr[left], arr[right]

        elif arr[left] + arr[right] < target:
            left += 1

        else:
            right -= 1

    return "Not possible"

twosum(arr, target)
```

```python
#Reverse an array.
from typing import List
arr = [21, 32, 3, -4, 0]

def reverseArray(arr: List[int]):

    left = 0 
    right = len(arr) -1 

    while left < right:
        
        arr[left],arr[right] = arr[right],  arr[left]

        left = left +1 
        right -=1

    return arr
reverseArray(arr)
```
