```python
# moveZeroes
from typing import List 
class Solution:
    def moveZeroes(self, nums: List[int]) -> None:
        """
        Do not return anything, modify nums in-place instead.
        """
        temp = []
        org = len(nums)

        for i in range(len(nums)):
            if nums[i] != 0:
                temp.append(nums[i])
            else:
                pass
        return temp + [0]*(org - len(temp))
        
nums = [0,1,0,3,12]
Solution().moveZeroes(nums)
```

```python
class Solution:
    def moveZeroes(self, nums: List[int]) -> None:
        """Do not return anything, modify nums in-place instead."""
        pos = 0

        for i in range(len(nums)):

            if nums[i] != 0:
                nums[pos] = nums[i]
                pos +=1

        #Fill remaining with zeros
        while pos < len(nums):
            nums[pos] = 0
            pos += 1
```

```python
# Input: nums = [-4,-1,0,3,10]
# Output: [0,1,9,16,100]
nums = [-4,-1,0,3,10]

class Solution:
    def sortedSquares(self, nums: List[int]) -> List[int]:
        
        out = []
        for i in range(len(nums)):
            out.append(nums[i]**2)

        return sorted(out)

Solution().sortedSquares(nums)     
```

```python
nums = [-1,0,1,2,-1,-4]
#Output: [[-1,-1,2],[-1,0,1]] Three Sum Zero

class Solution:
    def threeSum(self, nums: list[int]) -> list[list[int]]:
        nums.sort()
        print(nums)
        
        final = []

        for i in range(len(nums)):
            
            if i > 0 and nums[i] == nums[i-1]:
                continue

            left , right = i +1 , len(nums) - 1            
            val = nums[i]

            while left < right:
                print( f'i:{i}', left, right, f'SUM : {nums[left] + nums[right] + val}')
                
                if nums[left] + nums[right] + val ==0:
                    final.append([ nums[i], nums[left], nums[right] ])
                    left = left + 1
                    right = right -1

                    # Skip duplicate left values
                    while left < right and nums[left] == nums[left - 1]:
                        left += 1

                    # Skip duplicate right values
                    while left < right and nums[right] == nums[right + 1]:
                        right -= 1

                elif nums[left] + nums[right] + val > 0 :
                    right -= 1

                else:
                    left +=1
        return final

print(Solution().threeSum(nums))
```

```python
#backspaceCompare
s , t = "ab#c" , "ad#c"

#s, t  = "ab##", "c#d#"

def backspaceCompare( s: str, t: str) -> bool:

    final = ""

    for w in s:
        if w != '#':  
            final = final + w
        else:
            final = final[:-1]

    final2 = ""
    for w in t:
        if w != '#':  
            final2 = final2 + w
        else:
            final2 = final2[:-1]

    return final == final2


backspaceCompare( s, t)
```

```python
#14. Longest Common Prefix
strs = ["flight","flower","flow",]
    
def longestCommonPrefix( strs: List[str]) -> str:
    prefix = strs[0]
    
    for s in strs:    
        while not s.startswith(prefix):
                prefix = prefix[:-1]
    
    return prefix

longestCommonPrefix(strs)
```

```python
#Reverse an array.
from typing import List
arr = [21, 32, 3, -4, 0]

def reverseArray(arr: List[int]):

    left = 0 
    right = len(arr) -1 

    while left < right:
        
        arr[left],arr[right] = arr[right],  arr[left]

        left = left +1 
        right -=1

    return arr
reverseArray(arr)
```
