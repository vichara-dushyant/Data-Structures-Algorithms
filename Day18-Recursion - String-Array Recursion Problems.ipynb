```python
# a
# ab
# abbc
# abbcbccd

class Solution:
    def kthCharacter(self, k: int) -> str:
        
        def helper(k):
            if k ==1 :
                return 'a'

            # find smallest power of 2 >= k
            length = 1
            while length < k:
                length = length*2
            print(f'length : {length}')

            half = length // 2

            if k <= half:
                print(f'helper {k} running')
                return helper(k)
            else:
                ch = helper(k - half)
                print(f'helper {(k-half)} : {ch}')
                return chr(ord(ch) + 1)

        return helper(k)

k = 89
Solution().kthCharacter(k)
```

```python
digits = [1,2,3]
from typing import List 

class Solution:
    def find3digitsEvenNumbers(self, digits: List[int]) -> List[int]:

        s = set()
        # used = set()
        # path = []
        
        def helper(path, used):

            if len(path) == 3:
                if path[0] != 0 and path[2] % 2 == 0 :
                    num = path[0] * 100 + path[1] * 10 + path[2]
                    s.add(num)
                    print(f's : {s}')
                return

            for i in range(len(digits)):
                if i not in used:
                    path.append(digits[i])
                    used.add(i)
                    print(f'path, used : {path},{used}')

                    helper(path, used)

                    path.pop()
                    used.remove(i)
                    print(f'path poped, used removed i: {path},{used}')

        helper([], set())
        return sorted(list(s))
print(Solution().find3digitsEvenNumbers(digits))
```

```python
class Solution:
    def subsets(self, nums):

        def dfs(index):
            # Base case
            if index == len(nums):
                return [[]]

            subsets = dfs(index + 1)

            result = []

            for subset in subsets:
                # Exclude current element
                result.append(subset)

                # Include current element
                result.append([nums[index]] + subset)

            return result

        return dfs(0)

Solution().subsets(nums)
```

```python
from typing import List 

class Solution: #RecursiveBacktracking
    def subsets(self, nums):
        result = []

        def backtrack(index, subset):
            print(f"index={index}, subset={subset}")

            if index == len(nums):
                print("Adding:", subset)
                result.append(subset[:])
                return

            # Include
            subset.append(nums[index])
            print(f"Include {nums[index]}")
            backtrack(index + 1, subset)

            # Backtrack
            removed = subset.pop()
            print(f"Remove {removed}")

            # Exclude
            print(f"Exclude {nums[index]}")
            backtrack(index + 1, subset)

        backtrack(0, [])
        return result

Solution().subsets(nums)
```
