```python
class ListNode:
    def __init__(self, val, next = None):
        self.val = val
        self.next = next

# new_node = ListNode(2)
```

```python
class Stack:
    def __init__(self):
        self.top = None 

    def is_empty(self):
        return self.top == None

    def push(self, val):
        new_node = ListNode(val)
        new_node.next = self.top
        self.top = new_node
        
    def __str__(self):
        curr = self.top
        result = ''
        while curr is not None:
            result = result + str(curr.val) + "->" #+ "\n"
            #result = result + #"↓\n"
            curr = curr.next
        return result

    def pop(self):
        if self.top:
            data = self.top.val
            #print(f"removed self.top : {self.top.val}")
            self.top = self.top.next
            return data
        else:
            print("Empty stack")
            return "Empty stack?"

    def size(self):
        if self.is_empty():
            return "empty stack's size can't be determined"

        curr = self.top
        size = 0
        while curr:
            curr = curr.next
            size = size + 1
        return size

    
    def reverse_str_stack(self, text):
        s = Stack()
        for char in text:
            s.push(char)
        #print(s)
        
        res = ""
        while not s.is_empty():
            res = res + s.pop()
        return res

    def peek(self):
        if self.is_empty():
            return "Empty Stack"
        return self.top.val

```

```python
class Solution:
    def isValid(self, s: str) -> bool:
        l = []
        pairs = {'(': ')', '[': ']', '{': '}'}

        for ch in s:
            if ch in pairs:
                l.append(ch)
                print(l)
            else:
                if not l:
                    return False
                if pairs[l.pop()] != ch:
                    return False
        return not l

Solution().isValid("([])")
```

```python
class Solution:
    def makeGood(self, s: str) -> str:

        stack = []

        for ch in s:

            if (
                stack
                and stack[-1].lower() == ch.lower()
                and stack[-1] != ch
            ):
                stack.pop()
            else:
                stack.append(ch)

        return "".join(stack)

Solution().makeGood(s)
```

```python
import requests
from bs4 import BeautifulSoup

url = "https://docs.google.com/document/d/e/2PACX-1vSvM5gDlNvt7npYHhp_XfsJvuntUhq184By5xO_pA4b_gCWeXb6dM6ZxwN8rE6S4ghUsCj2VKR21oEP/pub"
response = requests.get(url)
soup = BeautifulSoup(response.text, 'html.parser')

grid = {}
for row in soup.find_all('tr')[1:]:
    cols = row.find_all('td')
    if len(cols) == 3:
        try:
            x = int(cols[0].get_text(strip=True))
            char = cols[1].get_text(strip=True)
            y = int(cols[2].get_text(strip=True))
            grid[(x, y)] = char
        except:
            continue

max_x = max(x for x, y in grid)
max_y = max(y for x, y in grid)
for y in range(max_y + 1):
    print("".join(grid.get((x, y), " ") for x in range(max_x + 1)))
```

```python
class MinStack:

    def __init__(self):
        self.stack = []
        self.smallest  = []

    def push(self, val: int) -> None:
        self.stack.append(val)
        print("Stack is :",self.stack)
        
        if self.smallest:    
            val = min(self.smallest[-1],val)
        self.smallest.append(val)
        print(self.smallest)

    def pop(self) -> None:
        self.stack.pop()
        self.smallest.pop()

    def top(self) -> int:
        return self.stack[-1]

    def getMin(self) -> int:
        return self.smallest[-1]
```
