```python
class PQ:

    maxHeap = []

    def swap(index, largestIndex):
        temp = PQ.maxHeap[index]
        PQ.maxHeap[index] = PQ.maxHeap[largestIndex]
        PQ.maxHeap[largestIndex] = temp

    def heapify(index):
        size = len(PQ.maxHeap)
        left = 2*index + 1
        right = 2*index + 2

        largestIndex = index

        if left < size and PQ.maxHeap[left]>PQ.maxHeap[largestIndex]:
            largestIndex = left

        if right < size and PQ.maxHeap[right]> PQ.maxHeap[largestIndex]:
            largestIndex = right 

        if largestIndex != index:
            PQ.swap(index, largestIndex)
            PQ.heapify(largestIndex)

    def insert(data):
        PQ.maxHeap.append(data)
        size = len(PQ.maxHeap)
        index = size - 1

        parent = (index-1)//2
        while parent >= 0 and PQ.maxHeap[index]>PQ.maxHeap[parent]:
            PQ.swap(index,parent)
            index = parent
            parent = (index -1 )//2

    def delete():
        size = len(PQ.maxHeap)
        if size == 0:
            return -1

        element = PQ.maxHeap[0]
        PQ.maxHeap[0] = PQ.maxHeap[size-1]
        PQ.maxHeap.pop()
        PQ.heapify(0)
        return element

    def print_heap():
        print(PQ.maxHeap)
```

```python
#Heapify Up (Min Heap)
def heapifyUp(arr, ind):

    while ind > 0:

        parent = (ind - 1) // 2

        if arr[ind] < arr[parent]:
            arr[ind], arr[parent] = arr[parent], arr[ind]
            ind = parent          # Continue from parent
        else:
            break                 # Heap property satisfied
```

```python
#Heapify Down (Min Heap)
def heapifyDown(arr, ind):

    n = len(arr)

    while True:

        smallest = ind

        left = 2 * ind + 1
        right = 2 * ind + 2

        if left < n and arr[left] < arr[smallest]:
            smallest = left

        if right < n and arr[right] < arr[smallest]:
            smallest = right

        if smallest == ind:
            break

        arr[ind], arr[smallest] = arr[smallest], arr[ind]
        ind = smallest
```
