```python
#https://leetcode.com/problems/assign-cookies/description/

def assigncookies(g,s): # 2 pointers (left for g, right for s) 
    # n = len(g) #total children
    # m = len(s)

    g.sort()   # O(n log n)
    s.sort()   # O(m log m)

    left = right = count = 0
    
    while left < len(g) and right < len(s):
        if g[left] <= s[right]:
            count += 1
            left += 1
        right += 1

    return count

#TC -> n logn + m log m + m
```

```python
#Fractional Knapsack https://www.geeksforgeeks.org/problems/fractional-knapsack-1587115620/1

val = [60, 100, 120, 40]
wt = [10, 20, 30, 13]
capacity = 50
#Output: 240

class Solution:
    def fractionalknapsack(self, val, wt, capacity):
        items = [(v/w, v, w) for v,w in zip(val,wt) if w>0] #avoid divide by 0
        items.sort(key = lambda x:x[0] , reverse = True)

        curr_wt = 0
        total_val = 0

        for ratio, v, w in items:

            if curr_wt + w <= capacity:
                #take the whole item
                total_val = total_val + v
                curr_wt = curr_wt + w

            else:
                #take only the fraction that fits
                remain = capacity - curr_wt
                
                if remain <= 0:
                    break
                else:
                    total_val = total_val + ratio*remain
                    break

        return total_val

Solution().fractionalknapsack(val, wt, capacity)
```

```python
# Minimum number of coins -> {1,2,5,10,50,100,200,500} - start loop from the end
#https://www.geeksforgeeks.org/problems/-minimum-number-of-coins4426/1
coins = [1,2,5,10,50]
n = len(coins)
res = []
num = 47

def findMin(num):
    for i in range(n-1, -1, -1):
        while num >= coins[i]:
            res.append(coins[i])
            num = num - coins[i]
        
    return res

#TC - O(n), SC - O(1)
findMin(num)
```

```python
# Why Greedy Works:
# Always pick the largest denomination possible first.
# Subtract it from the total and repeat the process.
# Continue until the total reduces to zero.
```

Intuition and Approach (Optimal – Greedy)
The greedy approach works perfectly here because the Indian currency system is structured such that larger denominations are multiples of smaller ones.

Why Greedy Works:
Always pick the largest denomination possible first.
Subtract it from the total and repeat the process.
Continue until the total reduces to zero.


```python
bills = [5,5,5,10,20,5]
#Output - true

def lemonadeChange(bills):
    n = len(bills)
    five = 0   # count of $5 bills
    ten = 0    # count of $10 bills
    
    for i in range(0, n):
        print(f'i:{i} {bills[i]} ,five:{five},ten:{ten}')
        if bills[i] == 5:
            five += 1
        elif bills[i] == 10:
            if five >= 1:
                five -= 1
                ten += 1
            else:
                return False
        else:  # customer pays with 20
            if ten >= 1 and five >= 1:
                # best case: give one $10 and one $5
                five -= 1
                ten -= 1
            elif five >= 3:
                # otherwise: give three $5 bills
                five -= 3
            else:
                return False
                
    return True

# Time Complexity:
# We only iterate through the list of customers once.
# O(n) where n = number of customers.
# Space Complexity:
# We only use two counters (five and ten).
# O(1) constant extra space.

lemonadeChange(bills)
```
