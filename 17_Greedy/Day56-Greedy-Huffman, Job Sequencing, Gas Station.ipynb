```python
start = [0, 3, 1, 6, 7, 11]
end   = [6, 5, 2, 8, 10, 15]
#        Y  N  N  N  Y   Y - 3 meetings
#           Y  Y  Y      Y - 4 meetings (Greedy because pick smallest meetings)
#The greedy approach works perfectly here because the best strategy is always to select the meeting that finishes earliest.
```

```python

class Meeting:
    def __init__(self, start, end, position):
        self.start = start
        self.end = end
        self.position = position

class Solution:
    def maxMeetings(self, start, end):
        n = len(start)

        # Create Meeting objects, Sort meetings by end time (then by start time if needed)
        meet = [Meeting(start[i], end[i], i+1) for i in range(n)]
        meet.sort(key = lambda x:(x.end, x.start))

        # start = [1, 3, 0, 6, 7, 11]
        # end   = [2, 5, 6, 8, 10, 15]
        
        lastTime = meet[0].end
        count = 1  # Select the first meeting by default

        for i in range(1,n):
            if meet[i].start > lastTime:
                count += 1
                lastTime = meet[i].end

        return count

Solution().maxMeetings(start,end)
```

```python
def activitySelection(start, finish):
    arr = list(zip(start, finish))
    arr.sort(key=lambda x: x[1])
    count = 1
    j = 0
    for i in range(1, len(arr)):
        # Check if current activity starts after last selected activity finishes
        if arr[i][0] > arr[j][1]:
            count += 1
            print(arr[i][0], ">" ,arr[j][1], "count = ", count)
            # Update last selected activity
            j = i
    return count
activitySelection(start,end)
```
