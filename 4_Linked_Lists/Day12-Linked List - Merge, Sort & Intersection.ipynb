```python
class ListNode:
    def __init__(self, val, next = None):
        self.val = val
        self.next = next
```

When you do: l1 = list1 you are not creating a copy of the linked list.
 You're creating another reference to the same nodes in memory.

curr = head, curr and head initially point to the same node, but when you do: curr = curr.next

you only move the curr reference; head still points to the original first node. Since you modify curr.val, you're changing the node objects themselves, so the changes are visible when you later return head.

```python
def bubble_sort(arr):
    n = len(arr)
    # Traverse through all array elements
    for i in range(n):
        # Last i elements are already in place, no need to re-check them
        for j in range(0, n - i - 1):

            print(i, j, n - i - 1, arr)
            # Swap if the element found is greater than the next element
            if arr[j] > arr[j + 1]:
                arr[j], arr[j + 1] = arr[j + 1], arr[j]

    return arr

bubble_sort([4,3,2,1])
```

```python
def merge_sort(arr):
    # Base case: an array of length 0 or 1 is already sorted
    if len(arr) <= 1:
        return arr

    # Step 1: Divide the array into two halves
    mid = len(arr) // 2
    left_half = arr[:mid]
    right_half = arr[mid:]

    # Step 2: Conquer by recursively sorting both halves
    left_sorted = merge_sort(left_half)
    right_sorted = merge_sort(right_half)

    # Step 3: Combine the sorted halves
    return merge(left_sorted, right_sorted)


def merge(left, right):
    sorted_arr = []
    i = 0  # Pointer for left array
    j = 0  # Pointer for right array

    # Compare elements from both arrays and merge them in order
    while i < len(left) and j < len(right):
        if left[i] < right[j]:
            sorted_arr.append(left[i])
            i += 1
        else:
            sorted_arr.append(right[j])
            j += 1

    # Append any remaining elements from the left or right list
    sorted_arr.extend(left[i:])
    sorted_arr.extend(right[j:])

    return sorted_arr


# Example Usage
if __name__ == "__main__":
    unsorted_array = [38, 27, 43, 3, 9, 82, 10]
    print(f"Unsorted: {unsorted_array}")

    result = merge_sort(unsorted_array)
    print(f"Sorted:   {result}")
```

```python
class Solution:
    def sortList(self, head: Optional[ListNode]) -> Optional[ListNode]:

        # Base case
        if not head or not head.next:
            return head

        # Find middle
        slow = head
        fast = head.next

        while fast and fast.next:
            slow = slow.next
            fast = fast.next.next

        # Split list into two halves
        mid = slow.next
        slow.next = None

        # Sort both halves
        left = self.sortList(head)
        right = self.sortList(mid)

        # Merge sorted halves
        return self.merge(left, right)

    def merge(self, l1, l2):

        dummy = ListNode(0)
        tail = dummy

        while l1 and l2:

            if l1.val <= l2.val:
                tail.next = l1
                l1 = l1.next
            else:
                tail.next = l2
                l2 = l2.next

            tail = tail.next

        if l1:
            tail.next = l1

        if l2:
            tail.next = l2

        return dummy.next
```
